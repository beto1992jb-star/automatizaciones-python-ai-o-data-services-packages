```python
import os
import json

# --- CONFIGURACIÓN SIMULADA DE API ---
# En un proyecto real, usaríamos las librerías oficiales (openai, anthropic, etc.)
# y las API Keys proporcionadas por el cliente. Para esta muestra, simulamos la respuesta.

def consultar_modelo_ia(texto_entrada, prompts):
    """
    Simula una llamada a un modelo de IA Generativa.
    Recibe el texto del usuario y las instrucciones (prompts).
    Retorna una respuesta estructurada en formato JSON.
    """
    print(f"--- Consultando modelo de IA con {prompts['modelo']} ---")
    print(f"PROMPT DE SISTEMA: {prompts['sistema']}")
    print(f"TEXTO DEL USUARIO: '{texto_entrada[:50]}...'")

    # Simulamos la respuesta del modelo basándonos en el texto de entrada.
    if "zapatillas" in texto_entrada.lower():
        respuesta_ia = {
            "titulo": "Zapatillas Deportivas SpeedMaster X1",
            "resumen": "Nuestras zapatillas SpeedMaster X1 ofrecen máxima amortiguación y diseño aerodinámico para corredores exigentes. Su tecnología de malla transpirable asegura frescura en cada kilómetro."
        }
    else:
        respuesta_ia = {
            "titulo": "Análisis General de Texto",
            "resumen": "No se detectó un producto específico, pero el texto fue procesado correctamente por la simulación de IA para el paquete básico."
        }

    return respuesta_ia

# --- FLUJO PRINCIPAL DEL SCRIPT ---

if __name__ == "__main__":
    # 1. Datos de entrada (lo que el cliente proporcionaría)
    texto_a_procesar = "Zapatillas deportivas de alto rendimiento para correr largas distancias. Incluye tecnología de amortiguación avanzada y diseño aerodinámico en color azul."

    # 2. Configuración del Prompt (lo que tú diseñas)
    config_prompt = {
        "modelo": "gpt-3.5-turbo (simulado)",
        "sistema": "Eres un asistente de marketing experto. Tu tarea es generar un título atractivo y un resumen corto de 2 oraciones para el producto descrito.",
        "temperatura": 0.5
    }

    print("\n--- INICIANDO PROCESAMIENTO BÁSICO DE IA ---\n")

    try:
        # 3. Llamada a la función (simulando la API)
        resultado = consultar_modelo_ia(texto_a_procesar, config_prompt)

        # 4. Mostrar y formatear el resultado
        print("\n--- RESULTADO DE LA IA (Formateado) ---\n")
        print(f"TÍTULO GENERADO: {resultado['titulo']}")
        print(f"RESUMEN GENERADO: {resultado['resumen']}")

        print("\n--- PROCESO COMPLETADO EXITOSAMENTE ---")

    except Exception as e:
        print(f"\n[ERROR] Ocurrió un problema durante el procesamiento: {e}")
```
